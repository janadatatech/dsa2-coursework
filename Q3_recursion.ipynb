## Coursework Question 3: Recursion (20 marks)

You must create a function that follows the following logic:

The function receives a list of strings. Each string can be either “blue”, “red” or “green”. Your goal is to move through the list incrementing the index by following the instructions below:

- If the string at index i is “blue”, the next element is the element following the next appearance of the string “blue”.
- If the string at index i is “red”, the next element is the element at index 2 * i + 1.
- If the string at index i is “green”, the next element is the following element in the list, i.e. at index i + 1.

The output should be a list with all the indexes of the elements you visited, in order.

### Example

For the list ["blue", "green", "red", "blue", "green", "red"], the sequence of steps would be:

1. The first step is the first element, “blue”, at index 0. Search for the next “blue” and take the following element. The next “blue” is at index 3, followed by “green” at index 4.
2. The second element is “green” at index 4. Go to the next element, which is “red” at index 5.
3. The third element is “red” at index 5. Go to index 2 * 5 + 1, which is 11. This index does not exist, so the program exits.

The output is [0, 4, 5].

### Tasks

a) Code the function using recursion. Answers without using recursion will score 0 for this question. Please use the following signature:

    def follow_instructions(instructions: list[str]) -> list[int]

i. Test it with 2 generated lists.

b) Create the flowchart of your code and put it in the report.

Need to code the function using recursion so start at index 0 and use a list to store the visited indexes.

Then use If statements check the colour 
if Green, move to i + 1.
if Red, move to 2 * i + 1.
if Blue, find the next blue and move to the position after it.

The function calls itself using the next index which is basically the recursion and it stops when the index is outside the list. 

If there is no next blue, also stop? Thats the assumption i will be working on

```python
def follow_instructions(instructions: list[str]) -> list[int]:
    visited = []

    def visit(index):
        if index >= len(instructions):
            return
        visited.append(index)
        colour = instructions[index]
        if colour == "green":
            visit(index + 1)
        elif colour == "red":
            visit(2 * index + 1)
        elif colour == "blue":
            for position in range(index + 1, len(instructions)):
                if instructions[position] == "blue":
                    visit(position + 1)
                    return

    visit(0)
    return visited
```

### Check example

I call the function using the example list from the question.

The expected result is [0, 4, 5] and I compare this with the returned list to check the result.

```python
example = ["blue", "green", "red", "blue", "green", "red"]
result = follow_instructions(example)
print("Visited indexes:", result)
print("Matches expected result:", result == [0, 4, 5])
```
